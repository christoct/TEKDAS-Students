## k-Nearest Neighbor

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
# Check data size
len(df_customer) #train_data
```

```python
# 3. Create the Model
model = KNeighborsClassifier(n_neighbors=7) # k = 7
```

```python
model
```

```python
X_train = df_customer.drop(['customer_id','churned'], axis=1)
y_train = df_customer['churned']
```

```python
# Train the model
model.fit(X_train, y_train)
```

```python
TARGET = "churned"
ID_COLUMN = "customer_id"

CATEGORICAL_COLUMNS = [
    "country",
    "gender",
    "membership_tier",
    "preferred_category",
    "preferred_device",
    "preferred_payment_method",
    "acquisition_channel",
]

NUMERIC_COLUMNS = [
    "age",
    "total_orders",
    "total_spend_usd",
    "avg_order_value_usd",
    "days_since_last_purchase",
    "reviews_given",
    "avg_review_score",
    "returns_made",
    "wishlist_items",
    "newsletter_subscribed",
    # engineered features
    "customer_tenure_days",
    "orders_per_year",
    "reviews_per_order",
    "returns_per_order",
]
```

```python
X_train = df_customer[NUMERIC_COLUMNS]
y_train = df_customer[TARGET]
```

```python
# Train the model
model.fit(X_train, y_train)
```

```python
# 4. Test the Model
X_test=df_customer_test[NUMERIC_COLUMNS]
y_test=df_customer_test[TARGET]
```

```python
y_pred = model.predict(X_test)
```

```python
result = pd.DataFrame({"y_test":y_test, "y_pred":y_pred})
result.head(10)
```

```python
# 5. Evaluation Metrics
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred, average="macro"))
print("Recall:", recall_score(y_test, y_pred, average="macro"))
print("F1 Score:", f1_score(y_test, y_pred, average="macro"))
```
